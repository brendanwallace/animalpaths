```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import math
import warnings

import util

warnings.filterwarnings("ignore")


FOLDER = "/Users/brendan/Projects/trails/data/"

# SIZE = 10000
STEPS = 20000

datafiles = [
    # "SVD|2026-01-09/data.json",
    "SVD-zoom|2026-05-25/data.json",
]
```

```python
dfs = [
    util.load_data(FOLDER + datafile) for datafile in datafiles
]
df = pd.concat(dfs)
util.describe(df)
df.columns
```

```python
# Perform the SVD
for seed in [1, 2]:
    for numLocations in [8, 16]:
        for p in ["SOLID", "PERIODIC"]:
            for cost in [2.0, 8.0]:
                for pR in [0.001, 0.0005]:
                    print(seed, numLocations, p, cost, pR)
                    _all = df[
                        (df["settings.patchRecovery"] == pR) &
                        (df["settings.numLocations"] == numLocations) &
                        (df["settings.randomSeedLocations"] == seed) &
                        (df["settings.boundaryConditions"] == p) &
                        (df["settings.maxCost"] == cost)
                    ].copy()
                
                
                    util.svd(_all, STEPS);
```
